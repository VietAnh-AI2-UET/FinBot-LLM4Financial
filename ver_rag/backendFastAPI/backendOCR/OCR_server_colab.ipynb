```python
from flask import Flask, request, jsonify
from flask_cors import CORS
from pyngrok import ngrok
import os, subprocess, threading, uuid
from datasets import load_dataset

app = Flask(__name__)
CORS(app)

UPLOAD_FOLDER = "/content/input"   # sửa bỏ khoảng trắng
OUTPUT_MD_FOLDER = "/content/ocr_results"

os.makedirs(UPLOAD_FOLDER, exist_ok=True)
os.makedirs(OUTPUT_MD_FOLDER, exist_ok=True)

# Bộ nhớ tạm lưu job status/result
jobs = {}
```

```python
def run_ocr_and_get_md(job_id):
    try:
        jobs[job_id] = {"status": "processing", "result": None}

        cmd = [
            "uv", "run", "nanonets-ocr.py",
            UPLOAD_FOLDER,
            "phuongkhanh123/my-ocr-output"
        ]
        subprocess.run(cmd, check=True)

        dataset = load_dataset("phuongkhanh123/my-ocr-output", split="train")
        md_contents = []
        for idx, example in enumerate(dataset):
            md_path = os.path.join(OUTPUT_MD_FOLDER, f"page_{idx+1:03}.md")
            with open(md_path, "w", encoding="utf-8") as f:
                f.write(example["markdown"])
            md_contents.append(example["markdown"])

        jobs[job_id] = {"status": "finished", "result": md_contents}
    except Exception as e:
        jobs[job_id] = {"status": "error", "result": str(e)}
```

```python
@app.route("/ocr", methods=["POST"])
def ocr_endpoint():
    files = request.files.getlist("files")
    if not files:
        return jsonify({"error": "No files uploaded"}), 400

    # Dọn thư mục input
    for filename in os.listdir(UPLOAD_FOLDER):
        os.remove(os.path.join(UPLOAD_FOLDER, filename))

    # Lưu file
    for i, file in enumerate(files):
        file.save(os.path.join(UPLOAD_FOLDER, f"page_{i+1:03}.png"))

    # Tạo job
    job_id = str(uuid.uuid4())
    t = threading.Thread(target=run_ocr_and_get_md, args=(job_id,))
    t.start()

    return jsonify({"job_id": job_id}), 202
```

```python
@app.route("/status/<job_id>", methods=["GET"])
def job_status(job_id):
    job = jobs.get(job_id)
    if not job:
        return jsonify({"error": "Job not found"}), 404
    return jsonify(job)
```

```python
if __name__ == "__main__":
    ngrok.kill()
    public_url = ngrok.connect(5555)
    print("Public URL:", public_url)
    app.run(port=5555)
```
